```python
from transformers import pipeline
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import matplotlib.pyplot as plt
```

```python
generator = pipeline(
    "text-generation",
    model="gpt2"
)

role = "Python Developer"
skills = "Python, Machine Learning, SQL"

prompt = f"""
Generate one interview question for a {role}
candidate having skills {skills}.
"""

question = generator(
    prompt,
    max_length=80,
    num_return_sequences=1
)[0]["generated_text"]

print("Interview Question:")
print(question)
```

```python
model = SentenceTransformer(
    'all-MiniLM-L6-v2'
)


candidate_answer = """
A list is mutable and can be changed.
A tuple is immutable and cannot be modified.
"""

model_answer = """
Lists are mutable data structures in Python.
Tuples are immutable and provide faster access.
"""


vectors = model.encode(
    [candidate_answer, model_answer]
)


similarity = cosine_similarity(
    [vectors[0]],
    [vectors[1]]
)[0][0]


score = int(similarity*100)


print("AI Evaluation")
print("----------------")
print("Score:",score)


if score > 80:
    print("Performance: Excellent")
elif score > 60:
    print("Performance: Good")
else:
    print("Performance: Needs Improvement")
```
